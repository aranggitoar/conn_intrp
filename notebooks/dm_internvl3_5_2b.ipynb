```python
run_dir = make_run_dir(WEIGHTS_PATH, "internvl", "dm")

run_dm(
    adapter, data_categorized,
    sparsity_coef=1.5e-3, lr=1.0, epochs=3, step=1,
    image_base_path=image_base_path, run_dir=run_dir,
)
```
